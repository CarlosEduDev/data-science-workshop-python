```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_netflix = pd.read_csv('NetFlix.csv')

df_netflix.head()

# df_netflix.info()

# print("Quantidade de valores faltantes nas colunas:")
# print(df_netflix.isnull().sum())
```

```python
df_netflix['country'] = df_netflix['country'].fillna('País não informado')
df_netflix['director'] = df_netflix['director'].fillna('Diretor não informado')
df_netflix['cast'] = df_netflix['cast'].fillna('Elenco não informado')
df_netflix['date_added'] = df_netflix['date_added'].fillna("Data não informada")
df_netflix['rating'] = df_netflix['rating'].fillna("Avaliação não informada")

apenas_filmes = df_netflix['type'] == 'Movie'

filmes = df_netflix[apenas_filmes].copy()

filmes.head()

# print("Total de filmes no catalogo: ", len(filmes))
```

# Operações vetoriais e Estatística Descritiva

```python
duracao = filmes['duration'].to_numpy()

media = np.mean(duracao)
mediana = np.median(duracao)
p90 = np.percentile(duracao, 90)
```

```python
plt.figure(figsize = (6, 4))
sns.countplot(data = df_netflix, x = 'type', palette = 'Set2')
plt.title("Distribuição de Catálogos da NetFlix")
plt.xlabel("Tipo de conteúdo")
plt.ylabel("Quantidade")
plt.show()
```
